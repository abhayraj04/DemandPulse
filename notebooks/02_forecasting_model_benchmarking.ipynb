# DemandVision AI — Forecasting Model Benchmarking
## Chronological Holdout Validation Across 5 Model Families

This notebook executes a walk-forward out-of-sample evaluation benchmarking:
1. **Seasonal Naive (Lag-7)**
2. **Moving Average (14-day)**
3. **Holt-Winters Exponential Smoothing (ETS)**
4. **ARIMA (1, 1, 1)**
5. **XGBoost Regressor (Lagged & Rolling Features)**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sys
from pathlib import Path

# Ensure backend module is importable
sys.path.append(str(Path.cwd().parent))
from backend.data.loader import get_data_loader
from backend.forecasting.validation import perform_walk_forward_validation

%matplotlib inline
```

### 1. Execute Walk-Forward Evaluation (Store 1, Item 1)

```python
loader = get_data_loader()
df_s1i1 = loader.get_data(store=1, item=1)
val_results = perform_walk_forward_validation(df_s1i1, test_horizon=30)

leaderboard_df = pd.DataFrame(val_results['leaderboard'])
print(f"Best Model: {val_results['best_model']}")
leaderboard_df
```

### 2. Plot Model Comparison (Actual vs Forecast on Holdout)

```python
comp_df = pd.DataFrame(val_results['comparison_points']).set_index('date')

fig, ax = plt.subplots(figsize=(14, 6))
comp_df['actual'].plot(ax=ax, color='black', lw=2.5, label='Actual Sales')
if 'XGBoost' in comp_df.columns:
    comp_df['XGBoost'].plot(ax=ax, color='#06B6D4', lw=2, label='XGBoost')
if 'ARIMA' in comp_df.columns:
    comp_df['ARIMA'].plot(ax=ax, color='#8B5CF6', lw=1.5, label='ARIMA')
if 'Naive' in comp_df.columns:
    comp_df['Naive'].plot(ax=ax, color='#94A3B8', linestyle='--', label='Seasonal Naive')

ax.set_title('30-Day Holdout Forecast vs Actual Observed Demand', fontsize=14, fontweight='bold')
ax.set_ylabel('Daily Units Sold')
ax.legend()
plt.tight_layout()
plt.show()
```
